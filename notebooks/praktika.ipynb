### Интерпретация результатов анализа корреляций с целевой переменной (yield)

Из полученного отсортированного списка коэффициентов корреляции Пирсона можно сделать  выводы:

1. Сильная положительная линейная связь (Драйверы урожайности):**
   * Наибольшее влияние на целевую переменную `yield` оказывают признаки `fruitset` (0.98), `seeds` (0.96) и `fruitmass` (0.93).  Чем лучше завязываются плоды, чем они крупнее и чем больше в них семян, тем выше итоговый сбор.  мультиколлинеарность между самими 

2. Умеренная положительная связь (Влияние опылителей):**
   * Среди насекомых-опылителей наибольшую линейную связь с урожайностью показывают одиночные пчелы `osmia` (0.38) и шмели `bumbles` (0.31). Дикие пчелы `andrena` (0.14) имеют слабую связь, а влияние домашних пчел `honeybee` (-0.04) стремится к нулю.

3. Отрицательная линейная связь (Лимитирующие факторы):**
   * Главными факторами, снижающими урожайность, выступают погодные условия: `AverageRainingDays` (-0.54) и `RainingDays` (-0.54), а также биологический фактор `clonesize` (-0.52). Осадки во время цветения явно ограничивают летную активность насекомых.

4. Слабая связь (Температурные диапазоны):**
   * Все показатели температурных диапазонов (группа признаков `Range`) сгруппировались в районе -0.18 ... -0.19. Они имеют стабильную, но довольно слабую отрицательную связь с урожайностью.
