```python
df_clean = df.drop(columns =['Row#'])
```

```python
#Групируем столбцы по категориям 
columns_by_groups = [
   'fruitset', 'fruitmass', 'seeds', # черника
   'clonesize', 'honeybee', 'bumbles','andrena', 'osmia', # пчелы и опыление 
   'MaxOfUpperTRange', 'MinOfUpperTRange', 'AverageOfUpperTRange', # Погода и дожди 
   'MaxOfLowerTRange', 'MinOfLowerTRange', 'AverageOfLowerTRange',
   'RainingDays', 'AverageRainingDays',
   'yield' # таргет 
]

df_ready = df_clean[columns_by_groups]
df_ready.to_csv ("../data/blueberry_clean.csv", index=False)

df_ready.head()
```

```python
# Строим общую матрицу корреляций что бы проверить связи признаков друг с другом  и найти дубли.
corr_matrix = df_ready.corr().round(2)
corr_matrix                            
```

```python
# список корреляций каждого признака с урожайностью по убыванию
df_ready.corr()['yield'].drop('yield').sort_values(ascending=False).round(2)
```

Из списка корреляций видно, какие факторы сильнее всего влияют на сбор ягод:

1. **Сильная прямая связь (что повышает урожай):**
   * Главные факторы успеха — это показатели самих плодов: завязываемость `fruitset` (`0.98`), количество семян `seeds` (`0.96`) и вес ягод `fruitmass` (`0.93`). Чем лучше завязываются и наливаются ягоды, тем выше итоговый урожай. Эти три показателя очень сильно связаны и друг с другом (почти дублируют смысл).

2. **Умеренная прямая связь (помощь опылителей):**
   * Из насекомых лучше всего помогают урожаю дикие пчёлы: осмии `osmia` (`0.38`) и шмели `bumbles` (`0.31`). Пчёлы андрены `andrena` (`0.14`) влияют слабо, а обычные домашние пчёлы `honeybee` (`-0.04`) на урожайность почти не влияют.

3. **Отрицательная связь (что снижает урожай):**
   * Сильнее всего мешают урожаю дожди: среднее число дождливых дней `AverageRainingDays` (`-0.54`) и общее количество дождливых дней `RainingDays` (`-0.54`), а также большой размер куста `clonesize` (`-0.52`). В дождливую погоду пчёлы хуже летают и меньше опыляют цветы.

4. **Слабая связь (температура):**
   * Все показатели температуры (признаки `Range`) дали слабую отрицательную связь на уровне `-0.18 ... -0.19`. Сами по себе перепады температур влияют на урожай незначительно.

```python
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.boxplot(df_ready['yield'])
plt.title('Выбросы в yield')

plt.subplot(1, 2, 2)
plt.boxplot(df_ready['RainingDays'])
plt.title('Выбросы в RainingDays')

plt.show()
```

На графике урожайности (yield) видны две низкие точки (ниже 2500), а в дождливых днях (RainingDays) выбросов нет. Поскольку эти низкие значения похожи на реальный неурожай, а не на сбой в таблице, я их оставила.

```python
plt.figure(figsize=(6, 4))
plt.scatter(df_ready['fruitset'], df_ready['yield'], color='coral', edgecolor='black', alpha=0.7)
plt.title('Зависимость yield от fruitset')
plt.xlabel('Завязываемость плодов (fruitset)')
plt.ylabel('Урожайность (yield)')

plt.tight_layout()
plt.show()
```

 Наблюдается сильная прямая зависимость между завязываемостью плодов (fruitset) и урожайностью (yield). С ростом показателя fruitset урожайность увеличивается практически линейно, что подтверждает высокий коэффициент корреляции (0.98).
