```python
df_clean = df.drop(columns =['Row#'])
```

```python
#Групируем столбцы по категориям 
columns_by_groups = [
   'fruitset', 'fruitmass', 'seeds', # черника
   'clonesize', 'honeybee', 'bumbles','andrena', 'osmia', # пчелы и опыление 
   'MaxOfUpperTRange', 'MinOfUpperTRange', 'AverageOfUpperTRange', # Погода и дожди 
   'MaxOfLowerTRange', 'MinOfLowerTRange', 'AverageOfLowerTRange',
   'RainingDays', 'AverageRainingDays',
   'yield' # таргет 
]

df_ready = df_clean[columns_by_groups]
df_ready.to_csv ("../data/blueberry_clean.csv", index=False)

df_ready.head()
```

```python
df_ready.describe()
```

```python
# Строим общую матрицу корреляций что бы проверить связи признаков друг с другом  и найти дубли.
corr_matrix = df_ready.corr().round(2)
corr_matrix                            
```

```python
corr_spearman = df_ready.corr(method='spearman').round(2)
corr_spearman
```

Вывод по корреляции Спирмена:
Эта таблица не боится выбросов, поэтому уточнила связь у медоносных пчёл (honeybee) с урожаем — вместо -0.04 стало -0.34. По всем остальным столбцам цифры почти не изменились.

```python
import seaborn as sns

plt.figure(figsize=(14, 10))
sns.heatmap(corr_spearman, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Тепловая карта корреляций (метод Спирмена)')
plt.tight_layout()
plt.show()
```

По цветам сразу видно, какие столбцы дублируют друг друга (ярко-красные квадраты 0.95–1.00): это все 6 столбцов с температурой и 3 показателя ягод (fruitset, fruitmass, seeds). В последнем столбце (yield) красным цветом выделено то, что повышает урожай (показатели ягод), а синим — то, что его снижает (дожди и большой размер куста).

```python
# список корреляций каждого признака с урожайностью по убыванию
df_ready.corr()['yield'].drop('yield').sort_values(ascending=False).round(2)
```

**Что видно из этого списка:**
* **Повышают урожай:** показатели самих ягод — завязываемость `fruitset` (`0.98`), семена `seeds` (`0.96`) и вес `fruitmass` (`0.93`). Чем их больше, тем лучше урожай.
* **Помогают пчёлы:** дикие пчёлы (осмии `0.38` и шмели `0.31`) идут в плюс. У домашних пчёл `honeybee` здесь почти ноль (`-0.04`), но это может быть из-за выбросов — проверим ниже.
* **Снижают урожай:** дожди (`-0.54`) и большой размер куста (`clonesize`, `-0.52`).
* **Почти не влияют:** температура (все столбцы около `-0.18`).

```python
plt.figure(figsize=(6, 4))
plt.scatter(df_ready['fruitset'], df_ready['yield'], color='coral', edgecolor='black', alpha=0.7)
plt.title('Зависимость yield от fruitset')
plt.xlabel('Завязываемость плодов (fruitset)')
plt.ylabel('Урожайность (yield)')

plt.tight_layout()
plt.show()
```
